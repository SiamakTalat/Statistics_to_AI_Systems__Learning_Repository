```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import poisson
np.random.seed(42)
```

## 1. Monte Carlo Estimation


```python
N = 5000
X = np.random.uniform(0, 1, N)
estimate = np.mean(X**2)
true_value = 1/3
pd.DataFrame({'Quantity':['Monte Carlo estimate','True value'], 'Value':[estimate, true_value]})
```

## 2. Simulation Error and Convergence


```python
running_est = np.cumsum(X**2) / np.arange(1, N+1)
plt.figure(figsize=(8,4))
plt.plot(running_est, label='Running estimate')
plt.axhline(true_value, linestyle='--', label='True value')
plt.title('Monte Carlo Convergence')
plt.xlabel('Number of samples')
plt.ylabel('Estimate of E[X^2]')
plt.legend()
plt.show()
```

```python
se = np.std(X**2, ddof=1) / np.sqrt(N)
pd.DataFrame({'Metric':['Monte Carlo standard error'], 'Value':[se]})
```

## 3. Random Walk Simulation


```python
steps = np.random.choice([-1, 1], size=300)
walk = np.cumsum(steps)
plt.figure(figsize=(8,4))
plt.plot(walk)
plt.title('1D Random Walk')
plt.xlabel('Time step')
plt.ylabel('Position')
plt.show()
```

## 4. Markov Chain Simulation


```python
P = np.array([[0.8, 0.2], [0.3, 0.7]])
states = [0]
for _ in range(99):
    current = states[-1]
    next_state = np.random.choice([0,1], p=P[current])
    states.append(next_state)
states = np.array(states)
pd.DataFrame({'State': [0,1], 'Empirical frequency': [(states==0).mean(), (states==1).mean()]})
```

```python
plt.figure(figsize=(8,3))
plt.step(np.arange(len(states)), states, where='post')
plt.title('Two-State Markov Chain')
plt.xlabel('Time step')
plt.ylabel('State')
plt.show()
```

## 5. Poisson Event Simulation


```python
lam = 4
counts = np.random.poisson(lam, size=1000)
pd.DataFrame({'Metric':['Sample mean','Sample variance'], 'Value':[counts.mean(), counts.var(ddof=1)]})
```

```python
plt.figure(figsize=(8,4))
plt.hist(counts, bins=np.arange(counts.min()-0.5, counts.max()+1.5, 1), density=True, alpha=0.7, label='Simulated')
k = np.arange(0, 15)
plt.plot(k, poisson.pmf(k, lam), marker='o', label='Poisson PMF')
plt.title('Poisson Event Counts')
plt.xlabel('Count')
plt.ylabel('Probability')
plt.legend()
plt.show()
```

## 6. Queue and Waiting-Time Simulation


```python
n_customers = 80
interarrival = np.random.exponential(scale=2.0, size=n_customers)
service = np.random.exponential(scale=1.5, size=n_customers)
arrival = np.cumsum(interarrival)
start = np.zeros(n_customers)
finish = np.zeros(n_customers)
wait = np.zeros(n_customers)
for i in range(n_customers):
    start[i] = arrival[i] if i == 0 else max(arrival[i], finish[i-1])
    wait[i] = start[i] - arrival[i]
    finish[i] = start[i] + service[i]
pd.DataFrame({'Metric':['Average waiting time','Average service time'], 'Value':[wait.mean(), service.mean()]})
```

```python
plt.figure(figsize=(8,4))
plt.plot(wait)
plt.title('Queue Waiting Times')
plt.xlabel('Customer')
plt.ylabel('Waiting time')
plt.show()
```

## 7. Brownian Motion


```python
T = 1.0
n_steps = 300
dt = T / n_steps
increments = np.random.normal(0, np.sqrt(dt), n_steps)
brownian = np.concatenate([[0], np.cumsum(increments)])
plt.figure(figsize=(8,4))
plt.plot(brownian)
plt.title('Brownian Motion Path')
plt.xlabel('Step')
plt.ylabel('Value')
plt.show()
```

## 8. Synthetic Data Generation


```python
n_syn = 200
x1 = np.random.normal(10, 2, n_syn)
x2 = np.random.normal(20, 4, n_syn)
y = 5 + 1.5*x1 - 0.7*x2 + np.random.normal(0, 2, n_syn)
syn_df = pd.DataFrame({'x1':x1, 'x2':x2, 'y':y})
syn_df.head()
```

## 9. Simulation-Based Decision Support


```python
N_dec = 5000
uncertain_demand = np.random.normal(100, 15, N_dec)
profit_A = 8 * uncertain_demand - 300
profit_B = 6 * uncertain_demand - 150
EU_A = profit_A.mean()
EU_B = profit_B.mean()
pd.DataFrame({'Decision':['A','B'], 'Expected Profit':[EU_A, EU_B]})
```

```python
summary = pd.DataFrame({
    'Measure':['MC estimate','MC standard error','Random walk final position','State 1 frequency','Mean Poisson count','Avg queue wait','Brownian final value','Best decision'],
    'Value':[estimate, se, walk[-1], (states==1).mean(), counts.mean(), wait.mean(), brownian[-1], 'A' if EU_A > EU_B else 'B']
})
summary
```
