```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
np.random.seed(42)
```

## 1. Expected Utility Decisions


```python
actions = ['A', 'B']
outcome_probs = np.array([0.6, 0.4])
utility_A = np.array([100, 20])
utility_B = np.array([75, 45])
EU_A = np.sum(outcome_probs * utility_A)
EU_B = np.sum(outcome_probs * utility_B)
pd.DataFrame({'Action': actions, 'Expected Utility': [EU_A, EU_B]})
```

## 2. Risk Aversion Simulation


```python
mean_return = np.array([10, 10])
variance = np.array([1, 9])
lam = 0.8
utility = mean_return - lam * variance
pd.DataFrame({'Option':['Low variance','High variance'], 'Mean':mean_return, 'Variance':variance, 'Mean-Variance Utility':utility})
```

## 3. Binary Choice Modeling Intuition


```python
n = 250
price = np.random.uniform(5, 25, n)
quality = np.random.uniform(1, 10, n)
latent = -0.35 * price + 0.55 * quality
prob_buy = 1 / (1 + np.exp(-latent))
buy = (np.random.rand(n) < prob_buy).astype(int)
X = np.column_stack([price, quality])
logit = LogisticRegression(max_iter=1000).fit(X, buy)
pred = logit.predict(X)
pd.DataFrame({'Feature':['price','quality'], 'Coefficient':logit.coef_[0]})
```

```python
pd.DataFrame({'Metric':['Accuracy','Intercept'], 'Value':[accuracy_score(buy, pred), logit.intercept_[0]]})
```

## 4. A/B Testing Example


```python
nA, nB = 500, 520
conv_A = np.random.binomial(1, 0.11, nA)
conv_B = np.random.binomial(1, 0.145, nB)
rate_A = conv_A.mean()
rate_B = conv_B.mean()
lift = rate_B - rate_A
z, p = stats.ttest_ind(conv_B, conv_A, equal_var=False)
pd.DataFrame({'Metric':['Rate A','Rate B','Lift (B-A)','p-value'], 'Value':[rate_A, rate_B, lift, p]})
```

## 5. Auction Bidding Simulation


```python
m = 200
valuation = np.random.uniform(50, 100, m)
shading = np.random.uniform(5, 20, m)
bid = valuation - shading
bid_df = pd.DataFrame({'valuation': valuation, 'shading': shading, 'bid': bid})
bid_df.head()
```

```python
plt.figure(figsize=(7,4))
plt.scatter(valuation, bid, s=12)
plt.title('Valuation vs Bid')
plt.xlabel('Valuation')
plt.ylabel('Bid')
plt.show()
```

## 6. Regret Analysis


```python
optimal_utility = 100
chosen_utilities = np.random.normal(85, 8, 100)
regret = optimal_utility - chosen_utilities
pd.DataFrame({'Metric':['Mean chosen utility','Mean regret','Max regret'], 'Value':[chosen_utilities.mean(), regret.mean(), regret.max()]})
```

```python
plt.figure(figsize=(7,4))
plt.hist(regret, bins=20)
plt.title('Regret Distribution')
plt.xlabel('Regret')
plt.ylabel('Frequency')
plt.show()
```

## 7. Sequential Decision Experiment Intuition


```python
gamma = 0.9
rewards = np.array([8, 7, 6, 5])
discounted_return = np.sum([gamma**k * rewards[k] for k in range(len(rewards))])
pd.DataFrame({'Metric':['Discount factor','Discounted return'], 'Value':[gamma, discounted_return]})
```

## 8. Summary Table


```python
summary = pd.DataFrame({
    'Measure':['Best expected utility action','Risk-averse preferred option','A/B lift','Average bid','Mean regret','Discounted return'],
    'Value':['A' if EU_A > EU_B else 'B', 'Low variance' if utility[0] > utility[1] else 'High variance', lift, bid.mean(), regret.mean(), discounted_return]
})
summary
```
